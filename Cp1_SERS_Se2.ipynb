# Dataset 1 — Appliances Energy Prediction

## Situação
Uma empresa de eficiência energética deseja identificar períodos de consumo elevado dos eletrodomésticos e observar as condições de temperatura nesses momentos.

```python
# 1 - Carregue a amostra e apresente head(), shape, info() e describe()
df1 = pd.read_csv('/content/Energy_DataSample.csv', skiprows=[1])

print("Primeiras linhas:")
display(df1.head())

print("\nDimensões:")
print(df1.shape)

print("\nInformações:")
df1.info()

print("\nEstatísticas:")
display(df1.describe())
```

```python
# 2 - Renomeie Appliances e pelo menos três atributos ambientais
df1.rename(columns={
    'Appliances': 'Consumo_Eletrodomesticos',
    'T1': 'Temperatura_1',
    'T2': 'Temperatura_2',
    'T3': 'Temperatura_3'
}, inplace=True)

df1.head()
```

```python
# 3 - Determine o maior consumo
PMAX = df1['Consumo_Eletrodomesticos'].max()
print(f"Maior consumo: {PMAX}")
```

```python
# 4 - Calcule 70% do máximo e crie o DataFrame acima desse limite
P70 = 0.70 * PMAX
print(f"Limiar de 70%: {P70}")

df2 = df1[df1['Consumo_Eletrodomesticos'] > P70]

df2.head()
```

```python
# 5 - Conte os registros selecionados e calcule o percentual da amostra
quantidade_registros = len(df2)
total_amostra = len(df1)

percentual = quantidade_registros / total_amostra * 100

print(f"Quantidade de registros: {quantidade_registros}")
print(f"Percentual da amostra: {percentual:.2f}%")
```

```python
# 6 - Calcule a temperatura média e aplique os dois critérios ao mesmo tempo
media_Temp1 = df1['Temperatura_1'].mean()

df_filtrado = df1[
    (df1['Consumo_Eletrodomesticos'] > P70) &
    (df1['Temperatura_1'] > media_Temp1)
]

print(f"Temperatura média de T1: {media_Temp1:.2f}")
df_filtrado.head()
```

```python
# 7 - Compare os dois DataFrames
print("Registros apenas pelo consumo:", len(df2))
print("Registros pelo consumo e temperatura:", len(df_filtrado))

# Resposta: a inclusão da temperatura como segundo critério torna o filtro
# mais restritivo, diminuindo a quantidade de registros selecionados.
```

# Dataset 2 — Steel Industry Energy Consumption

## Situação
Localizar situações de consumo elevado e verificar se coincidem com carga elevada ou fator de potência menos favorável.

```python
# 1 - Carregue a amostra
StIn = pd.read_csv('/content/Steel_Industry_DataSample.csv', skiprows=[1])

# Renomeie os principais atributos para nomes simples
StIn.rename(columns={
    'Usage_kWh': 'Consumo_kWh',
    'Lagging_Current_Power_Factor': 'Fator_Potencia_Atrasado',
    'Leading_Current_Power_Factor': 'Fator_Potencia_Adiantado'
}, inplace=True)

print("Primeiras linhas:")
display(StIn.head())

print("\nDimensões:")
print(StIn.shape)

print("\nInformações:")
StIn.info()

print("\nEstatísticas:")
display(StIn.describe())
```

```python
# 2 - Verifique as categorias de carga e os dias
print("Tipos de carga:")
print(StIn['Load_Type'].unique())

print("\nSituação da semana:")
print(StIn['WeekStatus'].unique())

print("\nValores ausentes:")
print(StIn.isnull().sum())
```

```python
# 3 - Determine o maior consumo e calcule 75% do máximo
PMAX_2 = StIn['Consumo_kWh'].max()
P75_2 = 0.75 * PMAX_2

print(f"Maior consumo: {PMAX_2:.2f} kWh")
print(f"Limiar de 75%: {P75_2:.2f} kWh")
```

```python
# 4 - Crie o DataFrame acima do limiar e calcule quantidade e percentual
df_alto_2 = StIn[StIn['Consumo_kWh'] > P75_2]

quantidade_2 = len(df_alto_2)
percentual_2 = quantidade_2 / len(StIn) * 100

print(f"Quantidade de registros: {quantidade_2}")
print(f"Percentual da amostra: {percentual_2:.2f}%")

df_alto_2.head()
```

```python
# 5 - Conte quantos registros de alto consumo são Maximum Load
maximum_load = df_alto_2[df_alto_2['Load_Type'] == 'Maximum Load']

print("Registros de alto consumo classificados como Maximum Load:", len(maximum_load))
```

```python
# 6 - Observe um fator de potência e defina um limite simples para valores baixos
print(StIn['Fator_Potencia_Atrasado'].describe())

# Resposta: vamos considerar 90% como limite.
# Valores abaixo de 90% serão tratados como fator de potência mais baixo.
limite_fp = 90
print("Limite escolhido para fator de potência:", limite_fp)
```

```python
# 7 - Crie um DataFrame com consumo elevado E fator de potência abaixo do limite
df_alto_fp = StIn[
    (StIn['Consumo_kWh'] > P75_2) &
    (StIn['Fator_Potencia_Atrasado'] < limite_fp)
]

print("Quantidade de registros:", len(df_alto_fp))
df_alto_fp.head()

# Resposta: esses registros merecem atenção porque combinam consumo elevado
# com fator de potência abaixo do limite escolhido.
```

```python
# 8 - Explique por que esse conjunto merece atenção
# Resposta:
# Um consumo elevado junto com fator de potência mais baixo pode indicar
# uma condição menos favorável para a operação elétrica da indústria.
# Por isso, esses registros podem ser analisados com mais atenção.
```

# Dataset 3 — Power Consumption of Tetouan City

## Situação
Identificar qual das três zonas apresenta o maior pico de consumo e observar a temperatura nos momentos de maior demanda.

```python
# 1 - Carregue a amostra
tetouan = pd.read_csv('/content/Tetuan_City_Power_Consumption_DataSample.csv', skiprows=[1])

# 2 - Renomeie as três variáveis de consumo
tetouan.rename(columns={
    'Zone 1 Power Consumption': 'Consumo_Zona_1',
    'Zone 2  Power Consumption': 'Consumo_Zona_2',
    'Zone 3  Power Consumption': 'Consumo_Zona_3'
}, inplace=True)

print("Primeiras linhas:")
display(tetouan.head())

print("\nDimensões:")
print(tetouan.shape)

print("\nInformações:")
tetouan.info()

print("\nEstatísticas:")
display(tetouan.describe())
```

```python
tetouan.columns
```

```python
# 3 - Determine o consumo máximo de cada zona
max_zona_1 = tetouan['Consumo_Zona_1'].max()
max_zona_2 = tetouan['Consumo_Zona_2'].max()
max_zona_3 = tetouan['Consumo_Zona_3'].max()

print("Máximo Zona 1:", max_zona_1)
print("Máximo Zona 2:", max_zona_2)
print("Máximo Zona 3:", max_zona_3)
```

```python
# 4 - Identifique a zona com maior pico
maior_pico = max(max_zona_1, max_zona_2, max_zona_3)

if maior_pico == max_zona_1:
    zona_maior = 'Consumo_Zona_1'
elif maior_pico == max_zona_2:
    zona_maior = 'Consumo_Zona_2'
else:
    zona_maior = 'Consumo_Zona_3'

print("Zona com maior pico:", zona_maior)
print("Maior pico:", maior_pico)
```

```python
# 5 - Calcule 70% do maior pico e filtre os registros acima desse limite
P70_tetouan = 0.70 * maior_pico

df_alto_tetouan = tetouan[tetouan[zona_maior] > P70_tetouan]

print(f"Limiar de 70%: {P70_tetouan:.2f}")
print("Quantidade de registros:", len(df_alto_tetouan))
```

```python
# 6 - Calcule o percentual dos registros selecionados
percentual_tetouan = len(df_alto_tetouan) / len(tetouan) * 100

print(f"Percentual da amostra: {percentual_tetouan:.2f}%")
```

```python
# 7 - Calcule a temperatura média e aplique os dois critérios
media_temp_tetouan = tetouan['Temperature'].mean()

df_alto_temp_tetouan = tetouan[
    (tetouan[zona_maior] > P70_tetouan) &
    (tetouan['Temperature'] > media_temp_tetouan)
]

print(f"Temperatura média: {media_temp_tetouan:.2f}")
print("Registros após os dois critérios:", len(df_alto_temp_tetouan))
```

```python
# 8 - Compare os dois conjuntos
print("Apenas consumo elevado:", len(df_alto_tetouan))
print("Consumo elevado + temperatura acima da média:", len(df_alto_temp_tetouan))

# Resposta: a quantidade diminui porque a temperatura acima da média
# foi adicionada como uma segunda condição.
```

# Dataset 4 — Solar Power Generation Data

## Situação
Localizar períodos de alta geração e identificar quais inversores aparecem com maior frequência nesses momentos.

```python
# 1 - Carregue a amostra
solar = pd.read_csv('/content/Sample_Plant_1_Generation_Data.csv', skiprows=[1])

# 2 - Renomeie as principais variáveis
solar.rename(columns={
    'DC_POWER': 'Potencia_CC',
    'AC_POWER': 'Potencia_CA',
    'DAILY_YIELD': 'Geracao_Diaria'
}, inplace=True)

print("Primeiras linhas:")
display(solar.head())

print("\nDimensões:")
print(solar.shape)

print("\nInformações:")
solar.info()

print("\nEstatísticas:")
display(solar.describe())
```

```python
solar.columns
```

```python
# 3 - Determine a maior potência CA
PCA_MAX = solar['Potencia_CA'].max()
print(f"Maior potência CA: {PCA_MAX:.2f}")
```

```python
# 4 - Calcule 70% da potência CA máxima
P70_solar = 0.70 * PCA_MAX
print(f"Limiar de 70%: {P70_solar:.2f}")
```

```python
# 5 - Crie o DataFrame de alta geração e calcule quantidade e percentual
df_alta_geracao = solar[solar['Potencia_CA'] > P70_solar]

quantidade_solar = len(df_alta_geracao)
percentual_solar = quantidade_solar / len(solar) * 100

print("Quantidade de registros:", quantidade_solar)
print(f"Percentual da amostra: {percentual_solar:.2f}%")
```

```python
# 6 - Conte os inversores nos registros de alta geração
contagem_inversores = df_alta_geracao['SOURCE_KEY'].value_counts()

print(contagem_inversores)
```

```python
# 7 - Identifique o inversor mais frequente
inversor_maior = contagem_inversores.idxmax()
quantidade_inversor = contagem_inversores.max()

print("Inversor mais frequente:", inversor_maior)
print("Quantidade de registros:", quantidade_inversor)

# Resposta: o inversor identificado aparece com maior frequência no recorte
# de alta geração. Isso permite observar sua presença nesses períodos,
# mas não permite concluir sozinho que ele tem melhor desempenho ou que
# os outros inversores apresentam falha.
```

# Dataset 5 — Wind & Solar Energy Production

## Situação
Comparar a ocorrência de períodos de alta produção solar e eólica, usando 70% do máximo de cada fonte.

```python
# 1 - Carregue a amostra
vento_solar = pd.read_csv('/content/Energy Production Dataset.csv', skiprows=[1])

# 2 - Renomeie as variáveis de produção
# No data set encontrado, não havia sentido trocar algumas colunas.
# Então as guardei em uma variável.

vento_solar.rename(columns={
    'Production': 'Produção'
}, inplace=True)

wind_data = vento_solar[vento_solar['Source'].str.contains('Wind', case=False, na=False)]

solar_data = vento_solar[vento_solar['Source'].str.contains('Solar', case=False, na=False)]

print("Primeiras linhas:")
display(vento_solar.head())

print("\nDimensões:")
print(vento_solar.shape)

print("\nInformações:")
vento_solar.info()

print("\nEstatísticas:")
display(vento_solar.describe())
```

```python
vento_solar.columns
```

```python
# 3 - Determine o máximo de cada fonte
max_solar = solar_data['Produção'].max()
max_eolica = wind_data['Produção'].max()

print("Máximo solar:", max_solar)
print("Máximo eólico:", max_eolica)
```

```python
# 4 - Calcule 70% do máximo de cada fonte
P70_solar = 0.70 * max_solar
P70_eolica = 0.70 * max_eolica

print("70% do máximo solar:", P70_solar)
print("70% do máximo eólico:", P70_eolica)
```

```python
# 5 - Crie os dois DataFrames
df_alto_solar = solar_data[solar_data['Produção'] > P70_solar]
df_alto_eolica = wind_data[wind_data['Produção'] > P70_eolica]

print("Alta geração solar:", len(df_alto_solar))
print("Alta geração eólica:", len(df_alto_eolica))
```

```python
# 6 - Calcule os percentuais e compare
percentual_alto_solar = len(df_alto_solar) / len(solar_data) * 100
percentual_alto_eolica = len(df_alto_eolica) / len(wind_data) * 100

print(f"Solar acima de 70%: {percentual_alto_solar:.2f}%")
print(f"Eólica acima de 70%: {percentual_alto_eolica:.2f}%")

if percentual_alto_solar > percentual_alto_eolica:
    print("A geração solar aparece com maior frequência acima de 70% do próprio máximo.")
elif percentual_alto_eolica > percentual_alto_solar:
    print("A geração eólica aparece com maior frequência acima de 70% do próprio máximo.")
else:
    print("As duas fontes aparecem com a mesma frequência.")
```

```python
# 7 - Explique por que os limites devem ser separados
# Resposta: solar e eólica possuem escalas e comportamentos diferentes.
# Por isso, usamos 70% do máximo de cada uma. Usar o mesmo valor numérico
# para as duas poderia favorecer uma fonte e prejudicar a comparação.
```

# Dataset 6 — Individual Household Electric Power Consumption

## Situação
Identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio.

```python
# 1 - Carregue a nova amostra
casa = pd.read_csv('/content/Household_DataSample (1).csv', skiprows=[1])

# Renomeie os atributos para nomes simples
casa.rename(columns={
    'Global_active_power': 'Potencia_Ativa',
    'Global_reactive_power': 'Potencia_Reativa',
    'Voltage': 'Tensao',
    'Global_intensity': 'Corrente',
    'Sub_metering_1': 'Submedicao_1',
    'Sub_metering_2': 'Submedicao_2',
    'Sub_metering_3': 'Submedicao_3'
}, inplace=True)

print("Primeiras linhas:")
display(casa.head())

print("\nDimensões:")
print(casa.shape)

print("\nInformações:")
casa.info()

print("\nEstatísticas:")
display(casa.describe())
```

```python
# 2 - Determine o valor máximo de potência ativa
PMAX_casa = casa['Potencia_Ativa'].max()
print(f"Maior potência ativa: {PMAX_casa}")
```

```python
# 3 - Calcule 75% do máximo e crie o DataFrame de demanda elevada
P75_casa = 0.75 * PMAX_casa

df_alta_potencia = casa[casa['Potencia_Ativa'] > P75_casa]

print(f"Limiar de 75%: {P75_casa:.2f}")
print("Quantidade de registros:", len(df_alta_potencia))
```

```python
# 4 - Calcule a quantidade e o percentual
quantidade_casa = len(df_alta_potencia)
percentual_casa = quantidade_casa / len(casa) * 100

print(f"Quantidade de registros: {quantidade_casa}")
print(f"Percentual da amostra: {percentual_casa:.2f}%")
```

```python
# 5 - Calcule a corrente média
media_corrente = casa['Corrente'].mean()
print(f"Corrente média: {media_corrente:.2f}")
```

```python
# 6 - Crie o segundo DataFrame com potência alta E corrente acima da média
df_alta_potencia_corrente = casa[
    (casa['Potencia_Ativa'] > P75_casa) &
    (casa['Corrente'] > media_corrente)
]

print("Quantidade de registros:", len(df_alta_potencia_corrente))
df_alta_potencia_corrente.head()
```

```python
# 7 - Compare os dois conjuntos
print("Apenas potência elevada:", len(df_alta_potencia))
print("Potência elevada + corrente acima da média:", len(df_alta_potencia_corrente))

# Resposta: a inclusão da corrente como segunda condição torna o filtro
# mais restritivo e reduz a quantidade de registros selecionados.
```
